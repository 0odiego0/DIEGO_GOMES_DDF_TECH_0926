```python
produtos = df.drop_duplicates("product_id")
```

```python
print(len(df), len(produtos), produtos["product_category_name"].nunique())
```

```python
def tirar(grupo):
    return grupo.sample(n=min(2, len(grupo)), random_state=42)

amostra = (
    produtos
    .groupby("product_category_name", group_keys=False)
    .apply(tirar)
    .reset_index(drop=True)
)

print(len(amostra), amostra["product_category_name"].nunique())
amostra[["product_id", "product_category_name", "product_category_name_english", "price"]].head(10)
```

```python
import time

def gerar_ficha(produto):
    pedido_texto = f"""
Escreva a ficha de um produto de marketplace brasileiro.
Gere um título curto e uma descrição de 4 a 6 frases, em português.
Mencione o preço em reais. Não invente marca.
Atributos:
- categoria: {produto.product_category_name}
- preço em reais: {produto.price}
- peso em gramas: {produto.product_weight_g}
- dimensões em cm (C x A x L): {produto.product_length_cm} x {produto.product_height_cm} x {produto.product_width_cm}
Responda apenas neste JSON:
{{"titulo": "...", "descricao": "..."}}
"""
    texto = client.chat.completions.create(
        model="gpt-4o-mini",
        response_format={"type": "json_object"},
        messages=[{"role": "user", "content": pedido_texto}],
        temperature=0.4,
    )
    ficha = json.loads(texto.choices[0].message.content)

    pedido_features = f"""
Extraia features somente do título e da descrição.
category: categoria curta extraída do texto.
price_tier: apenas baixo, medio ou alto.
baixo se o preço mencionado for menor que 50, medio de 50 a 200, alto acima de 200.
Se material, público, cor ou compatibilidade não estiverem no texto, use null.
Título: {ficha["titulo"]}
Descrição: {ficha["descricao"]}
Responda apenas neste JSON:
{{
  "category": "",
  "material": null,
  "audience": null,
  "features": {{"use_case": "", "price_tier": ""}},
  "color_options": [],
  "compatibility": null
}}
"""
    extraido = client.chat.completions.create(
        model="gpt-4o-mini",
        response_format={"type": "json_object"},
        messages=[{"role": "user", "content": pedido_features}],
        temperature=0,
    )
    features = json.loads(extraido.choices[0].message.content)
    return {
        "product_id": produto.product_id,
        "product_category_name": produto.product_category_name,
        "price": produto.price,
        "titulo": ficha["titulo"],
        "descricao": ficha["descricao"],
        "features_json": json.dumps(features, ensure_ascii=False),
    }

linhas = []
for i, produto in amostra.iterrows():
    try:
        linhas.append(gerar_ficha(produto))
    except Exception as erro:
        linhas.append({"product_id": produto.product_id, "erro": str(erro)})
    if len(linhas) % 10 == 0:
        print(len(linhas), "de", len(amostra))
    time.sleep(0.2)

resultado = pd.DataFrame(linhas)
resultado.to_csv("features_amostra.csv", index=False)
resultado.head(3)
```
